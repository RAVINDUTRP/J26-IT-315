```python
# =========================================================================
# 1. ENVIRONMENT SETUP & LINKED XGBOOST ARTIFACT LOADING
# Linked Module: src/model_training.py (XGBoostModelTrainer)
# =========================================================================
import sys
import json
from pathlib import Path
import pandas as pd
import joblib

cur = Path.cwd().resolve()
root_dir = cur if (cur / 'config' / 'project_config.json').exists() else cur.parent
if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

xgb_model = joblib.load(root_dir / 'models' / 'xgboost' / 'xgboost_model.joblib')
with open(root_dir / 'models' / 'xgboost' / 'xgboost_metadata.json', 'r') as f:
    meta = json.load(f)

print('XGBoost Model Architecture & Hyperparameters:')
print(f'Best Iteration: {meta["best_iteration"]}')
print(f'Scale Pos Weight: {meta["scale_pos_weight"]}')

print('\nTop 10 Most Influential Features (Gain):')
pd.Series(meta['top_10_features'])
```

### XGBoost Performance Analysis
XGBoost achieves exceptional precision (99.8%) and a false-positive rate near zero (0.0000) on the untouched chronological test set. Top predictive drivers include rolling 3h/6h rainfall totals, turbidity lag, and rate-of-change indicators.
