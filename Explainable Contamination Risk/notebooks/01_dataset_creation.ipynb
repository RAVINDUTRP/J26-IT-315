# Notebook 01 — Deterministic Environmental Time-Series Dataset Generation
**Component 03: Hybrid Explainable Contamination Risk Prediction Framework**

### Research Purpose
This notebook executes deterministic, hydrodynamic-coupled synthetic time-series generation for 4 river monitoring sites. It simulates coupled physical dynamics between rainfall, delayed runoff, turbidity surges, dilution or industrial spikes in TDS/EC, and historical contamination events.

> **Scientific Integrity Notice**:
> This dataset is synthetically generated under controlled hydrodynamic assumptions for algorithm design and benchmarking. > It does **not** represent measured real-world pollution records from Sri Lankan waterways. Performance measured on this dataset > verifies algorithmic learning capacity, not field validation accuracy.

```python
# =========================================================================
# 1. ENVIRONMENT BOOTSTRAP & LINKED MODULE IMPORTS
# Linked Modules: src/utils.py, src/data_generator.py, src/data_loader.py
# =========================================================================
import sys
from pathlib import Path
import pandas as pd
import numpy as np

# Robust project root discovery
cur = Path.cwd().resolve()
root_dir = cur if (cur / 'config' / 'project_config.json').exists() else cur.parent
if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

# Import core utilities and generators from src/
from src.utils import get_project_root, load_project_config
from src.data_generator import generate_synthetic_dataset
from src.data_loader import load_raw_data

print(f'Project Root: {get_project_root()}')
```

```python
# =========================================================================
# 2. EXECUTE DATASET GENERATION / LOAD CACHED RAW OBSERVATIONS
# Linked Function: generate_synthetic_dataset() from src/data_generator.py
# =========================================================================
try:
    # Load raw observations if already generated on disk
    df_wq = load_raw_data()
    print('Loaded existing raw water quality dataset from data/raw/!')
except Exception:
    # Otherwise generate deterministically using fixed random seed = 42
    df_wq, _ = generate_synthetic_dataset(seed=42)
    print('Generated new synthetic dataset!')

print(f'Total observations: {len(df_wq):,}')
```

```python
# =========================================================================
# 3. DATASET PROVENANCE AUDIT & CLASS DISTRIBUTION INSPECTION
# =========================================================================
print('Dataset Columns:')
print(df_wq.columns.tolist())

print('\nTarget Class Balance (future_contamination_event):')
print(df_wq['future_contamination_event'].value_counts(normalize=True).apply(lambda x: f'{x:.2%}'))

print('\nRisk Category Breakdown:')
print(df_wq['future_risk_category'].value_counts(normalize=True).apply(lambda x: f'{x:.2%}'))
df_wq.head()
```
